## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

I picked lane 2 which is about refresh/content opportunity scoring. The reason I picked this lane is because I believe it creates action items for the client/content editors rather than just observations. By creating a score that combines both decline severity and probability of recovery, subject matter experts would have recommendations on which pages to prioritize for reviewing.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

A real issue that this could solve is the limited time that editors have to review pages. The work could create a ranked queue based on the final scores to use their time more effectively. Ultimately, it would be up to the subject matter expert for final human input to cross-check the score with the page to decide which action to take.

The cost of a false positive (high rank but is worth fixing) is the direct cost of wasting time which is a limited resource in this scenario.
The cost of a false negative (low rank but is worth fixing) is the opportunity cost of lost traffic. I would weigh both false positive and false negative costs equally because both editor time and lost traffic seem to be equally undesirable losses to a business.

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")  # move from notebooks/ to the repo root

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — are you at the repo root?"
print("Starter data found. You're ready.")

import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
```

```python
print(f"1. Declining pages with real demand")
declining_with_demand = df[(df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)]
pct_declining_with_demand = len(declining_with_demand) * 100.0 / len(df)
impressions_at_stake = declining_with_demand["impressions_90d"].sum()
clicks_at_stake = declining_with_demand["clicks_90d"].sum()

print(f"{len(declining_with_demand):,} of {len(df):,} pages ({pct_declining_with_demand:.2f}%)")
print(f"Impressions at stake: {impressions_at_stake}")
print(f"Clicks at stake: {clicks_at_stake}")
```

```python
print(f"2. Top-10 ranked pages that are quietly declining")
page_top = df[df["avg_position"].between(1, 10)]
page_top_declining = page_top[page_top["trend_direction"] == "down"]
pct_page_top_declining = len(page_top_declining) * 100.0 / len(page_top)
impressions_at_stake = page_top_declining["impressions_90d"].sum()
clicks_at_stake = page_top_declining["clicks_90d"].sum()

print(f"{len(page_top_declining):,} of {len(page_top):,} top pages ({pct_page_top_declining:.1f}%)")
print(f"Impressions at stake: {impressions_at_stake}")
print(f"Clicks at stake: {clicks_at_stake}")
```
